```python

import selenium
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time
from sqlalchemy import create_engine
import mysql.connector
import glob
import os
from selenium.webdriver.common.keys import Keys



driver = webdriver.Chrome()


#overall URL
#driver.get("https://www.imdb.com/search/title/?title_type=feature&release_date=2024-01-01,2024-12-31")

#aAdventure genre
#driver.get("https://www.imdb.com/search/title/?title_type=feature&release_date=2024-01-01,2024-12-31&genres=adventure")
#Comedy genre
#driver.get("https://www.imdb.com/search/title/?title_type=feature&release_date=2024-01-01,2024-12-31&genres=comedy")
#Fantasy genre
#driver.get("https://www.imdb.com/search/title/?title_type=feature&release_date=2024-01-01,2024-12-31&genres=fantasy")
#Crime genre
driver.get("https://www.imdb.com/search/title/?title_type=feature&release_date=2024-01-01,2024-12-31&genres=crime")

Movie_Name = []
Genre =[]
Voting_Count = []
Rating =[]
Duration = []

#Total content XPATH
#movie_items =driver.find_elements(By.XPATH,'//*[@id="__next"]/main/div[2]/div[3]/section/section/div/section/section/div[2]/div/section/div[2]/div[2]/ul/li')

movie_items =driver.find_elements(By.XPATH,'//*[@id="__next"]/main/div[2]/div[3]/section/section/div/section/section/div[2]/div/section/div[2]/div[2]/ul/li')


for i in movie_items:
  Movie_Title = i.find_element(By.XPATH,'./div/div/div/div[1]/div[2]/div[1]/a/h3').text
  Genre_Details = i.find_element(By.XPATH,'//*[@id="__next"]/main/div[2]/div[3]/section/section/div/section/section/div[2]/div/section/div[1]/div/div/div[2]/button[3]').text
  #Genre_Details = i.find_element(By.XPATH,'./div[1]/div/div/div[2]/button[3]').text

  rating = i.find_element(By.XPATH, './div/div/div/div[1]/div[2]/span/div/span/span[1]').text
  voting = i.find_element(By.XPATH, './div/div/div/div[1]/div[2]/span/div/span/span[2]').text
  duration = i.find_element(By.XPATH, './div/div/div/div[1]/div[2]/div[2]/span[2]').text
  Movie_Name.append(Movie_Title)
  Genre.append(Genre_Details)
  Rating.append(rating)
  Voting_Count.append(voting)
  Duration.append(duration)

#print(Movie_Name , Voting_Count, Rating,Duration)
df = pd.DataFrame({
   'Title': Movie_Name,
   'Genre' : Genre,
  'Rating': Rating,
 'Votes': Voting_Count,
'Duration': Duration
})


#Genre1 -Adventure 
df.to_csv('Crime.csv', index=False)
df = pd.read_csv('Crime.csv')
print(df)
# #########################################
# #Genre2 - Comedy
#df.to_csv('Comedy.csv', index=False)
#df = pd.read_csv('Comedy.csv')
#print(df)
# ##########################################
# #Genre3- fantasy 

#df.to_csv('fantasy.csv', index=False)
#df3 = pd.read_csv('fantasy.csv')
# print(df3)
# ##########################################
# #Genre4-Family
# df.to_csv('Family.csv', index=False)
# df6 = pd.read_csv('Family.csv')
# print(df6)
# #############################



#Merging Dataframes.


# path = "C:/Users/Deiva/Desktop/Deiva_Project/Final"

# all_files = glob.glob(os.path.join(path, "*.csv"))

# # Initialize an empty list to hold DataFrames
# df_list = []

# Loop through the list of files and read each one into a DataFrame
# for file in all_files:
#     df = pd.read_csv(file)
#     df_list.append(df)

# Concatenate all DataFrames in the list into a single DataFrame
#merged_df = pd.concat(df_list, ignore_index=False)
#merged_df.to_csv("C:/Users/Deiva/Desktop/Deiva_Project/Final/merged_file.csv", index=False)




##############################################################
```
